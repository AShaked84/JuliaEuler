```julia
function largest_prime_factor(my_num)
    num = my_num
    factor = 2

    while factor * factor <= num
        num%factor == 0 ? num = num ÷ factor : factor == 2 ? factor += 1 : factor += 2
    end

    return num
end
```

```julia
num = 600851475143
print("Largest prime factor: ", largest_prime_factor(num))
```
