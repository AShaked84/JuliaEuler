```julia
function seive(max_num)
    primes = BitVector(undef, max_num)
    fill!(primes, true)
    i = 2

    primes[1] = false

    while i*i <= max_num
        if primes[i] == 1
            for j in i*i:i:max_num
                primes[j] = 0
            end
        end
        i+=1
    end
    return primes
end
```

```julia
max_prime = 2000000
sum = 0

my_primes = seive(max_prime)
s = 0
for i in 2:max_prime
    my_primes[i] && (s += i)
end
println(s)
```
