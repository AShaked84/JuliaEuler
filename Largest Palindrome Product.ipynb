```julia
function LargestPalindromeProduct(maxNum)
    largest = 0

    for i in maxNum:-1:10
        for j in maxNum:-1:10
            num_string = string(i*j)
            if num_string == reverse(num_string) && i*j > largest
                largest = i*j
            end
        end
    end

    return largest
end
```

```julia
print(LargestPalindromeProduct(99))
print("\n")
print(LargestPalindromeProduct(999))
```
