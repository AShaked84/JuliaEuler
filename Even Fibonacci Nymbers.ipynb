```julia
function EvenFibonacciSum(maxNum)
    sum = 0
    a = 1
    b = 2

    while a <= maxNum
        if a%2 == 0 
            sum +=a
        end
        
        temp = a + b
        a = b
        b = temp
    end

    return sum
end
```

```julia
print(EvenFibonacciSum(90))
print("\n"
print(EvenFibonacciSum(4000000))
```
