```julia
function GCD(a, b)
    while b != 0
        temp = b
        b = a%b
        a = temp
    end
    return a
end
```

```julia
function smallest_multiple(number_list)
    lcm = 1

    for i in number_list
        gcd = GCD(i, lcm)
        lcm = (lcm * i) ÷ gcd
    end

    return lcm
end
```

```julia
print(smallest_multiple(1:10))
print("\n")
print(smallest_multiple(1:20))
```
